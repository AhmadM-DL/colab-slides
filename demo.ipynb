# colab-slides demo
Paste a slide link and run the cell to see the deck below it.

```
!pip install -q colab-slides
```

```
%load_ext colab_slides
```

## 1. Google Slides link (embedded viewer, no install)

```
%slides https://docs.google.com/presentation/d/YOUR_DECK_ID/edit
```

## 2. Direct .pptx link
`auto` embeds through Office Online; add `--local` to render inside Colab.

```
url = "https://example.com/your-deck.pptx"  #@param {type:"string"}
%slides $url --local
```

## 3. Private deck in your Google Drive

```
from google.colab import drive
drive.mount('/content/drive')
%slides /content/drive/MyDrive/deck.pptx
```

## 4. Upload from your computer

```
from google.colab import files
up = files.upload()
import colab_slides
colab_slides.show(next(iter(up)), mode="local")
```
